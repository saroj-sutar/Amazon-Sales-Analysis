```python
df["TotalAmount"] = df["TotalAmount"].fillna(((df["Quantity"]*df["UnitPrice"])-((df["Quantity"]*df["UnitPrice"])*df["Discount"]))+
                                             (df["Tax"]+df["ShippingCost"]))
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df["OrderID"].duplicated().sum()
```

```python
df.columns
```

```python
for i in df.columns:
    if df[i].nunique() < 50:
        print(df[i].value_counts())
        print("-"*50)
```

```python
df["PaymentMethod"] = df["PaymentMethod"].str.capitalize()
```

```python
df["PaymentMethod"].value_counts()
```

```python
df["PaymentMethod"] = df["PaymentMethod"].replace({"Upi" : "UPI", "Cod" : "Cash on delivery", "Ap" : "Amazon pay","Cc" : "Credit card"})
```

```python
df["PaymentMethod"].value_counts()
```

```python
df["Country"] = df["Country"].str.capitalize()
```

```python
df["Country"].value_counts()
```

```python
df["Country"] = df["Country"].replace({"Us" : "United states", "Uk" : "United Kingdom"})
```

```python
df["Country"].value_counts()
```

```python
df.describe()
```

```python
display(df.describe(include="all").T)
```

```python
print("1. Missing values:")
print(df.isnull().sum())

print("\n2. Total missing values:")
print(df.isnull().sum().sum())

print("\n3. Duplicate rows:")
print(df.duplicated().sum())

print("\n4. Duplicate Order IDs:")
print(df["OrderID"].duplicated().sum())

print("\n5. Payment method values:")
print(df["PaymentMethod"].value_counts())

print("\n6. Country values:")
print(df["Country"].value_counts())

print("\n7. OrderDate data type:")
print(df["OrderDate"].dtype)
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
```

```python
df["OrderDate"].dtype
```

```python
df.columns
```

```python
# Which product category generates the highest total sales?
df.groupby("Category")["TotalAmount"].sum().round().sort_values(ascending = False).head(1)
```

```python
# Which payment method generates the highest total sales?
df.groupby("PaymentMethod")["TotalAmount"].sum().round().sort_values(ascending = False).head(1)
```

```python
# Which countries generate the highest total sales?
df.groupby("Country")["TotalAmount"].sum().round().sort_values(ascending = False).head(1)
```

```python
# Which product brands generate the highest total sales?
df.groupby("Brand")["TotalAmount"].sum().round().sort_values(ascending = False).head(1)
```

```python
# Which cities generate the highest total sales?
df.groupby("City")["TotalAmount"].sum().round().sort_values(ascending = False).head(1)
```

```python
#Which states generate the highest total sales?
df.groupby("State")["TotalAmount"].sum().round().sort_values(ascending = False).head(1)
```

```python
# How do total sales change month by month?
df.groupby(df["OrderDate"].dt.to_period("M"))["TotalAmount"].sum().round().sort_index()
```

```python
# Total sales by order status
df.groupby("OrderStatus")["TotalAmount"].sum().round()
```

```python
# Top 10 products by total sales
df.groupby("ProductName")["TotalAmount"].sum().round().sort_values(ascending = False).head(10)
```

```python
# Top 10 customers by total sales
df.groupby("CustomerName")["TotalAmount"].sum().round().sort_values(ascending = False).head(10)
```

```python
# Average order amount by product category
df.groupby("Category")["TotalAmount"].mean().round(2)
```

```python
# Total quantity sold by product category
df.groupby("Category")["Quantity"].sum().round(2)
```

```python
# Average discount by product category
df.groupby("Category")["Discount"].mean()
```

```python
# Average unit price by product category
df.groupby("Category")["UnitPrice"].sum()
```

```python
# Total tax by product category
df.groupby("Category")["Tax"].sum()
```

```python
# Total shipping cost by product category
df.groupby("Category")["ShippingCost"].sum()
```

```python
# Number of orders by payment method
df.groupby("PaymentMethod")["OrderID"].count()
```

```python
# Which brands have the highest total quantity of products sold?
df.groupby("Brand")["Quantity"].sum().sort_values(ascending = False).head(1)
```

```python
# How does the number of unique orders change month by month?
df.groupby(df["OrderDate"].dt.to_period("M"))["OrderID"].count().sort_index()
```
